```python
import pandas as pd
import numpy as np

df = pd.read_csv("/content/Crop_data - Sheet1 (1).csv", header=0)
df = df.dropna(how='all')


germination = {
    "Control": df["Control_Shoot_Length"].count(),
    "T1": df["T1_Shoot_Length"].count(),
    "T2": df["T2_Shoot_Length"].count()
}

## Agronomic_Indices
def calculate_agronomic_metrics(shoot_col, root_col, germ_percent):

    mean_shoot = shoot_col.mean()
    mean_root = root_col.mean()

    std_shoot = shoot_col.std()
    std_root = root_col.std()

    cv_shoot = (std_shoot / mean_shoot) * 100
    cv_root = (std_root / mean_root) * 100

    vigor_index = germ_percent * (mean_shoot + mean_root)

    return {
        "Mean_Shoot_Length": round(mean_shoot, 2),
        "Mean_Root_Length": round(mean_root, 2),
        "SD_Shoot": round(std_shoot, 2),
        "SD_Root": round(std_root, 2),
        "CV_Shoot_%": round(cv_shoot, 2),
        "CV_Root_%": round(cv_root, 2),
        "Seedling_Vigor_Index": round(vigor_index, 2)
    }

## Metrics for Each Treatment
results = pd.DataFrame({
    "Control": calculate_agronomic_metrics(
        df["Control_Shoot_Length"],
        df["Control_Root_Length"],
        germination["Control"]
    ),
    "T1": calculate_agronomic_metrics(
        df["T1_Shoot_Length"],
        df["T1_Root_Length"],
        germination["T1"]
    ),
    "T2": calculate_agronomic_metrics(
        df["T2_Shoot_Length"],
        df["T2_Root_Length"],
        germination["T2"]
    )
}).T  # Transpose for better view

print(results)
```

Part 2 Independent Statistic

```python
from scipy import stats
```

```python
for col in ["Control_Shoot_Length", "T1_Shoot_Length", "T2_Shoot_Length"]:
    print(f"\nColumn: {col}")
    print("Datatype:", df[col].dtype)
    print("Missing values:", df[col].isna().sum())
    print("Count:", df[col].count())
    print("Std Dev:", df[col].std())
```

```python
t_stat_t1_shoot, p_value_t1_shoot = stats.ttest_ind(
    df["Control_Shoot_Length"],
    df["T1_Shoot_Length"],
    equal_var=False,
    nan_policy='omit'
)

t_stat_t2_shoot, p_value_t2_shoot = stats.ttest_ind(
    df["Control_Shoot_Length"],
    df["T2_Shoot_Length"],
    equal_var=False,
    nan_policy='omit'
)

print("Shoot Length Comparison")
print("Control vs T1 → t-stat:", round(t_stat_t1_shoot, 3),
      "| p-value:", round(p_value_t1_shoot, 5))

print("Control vs T2 → t-stat:", round(t_stat_t2_shoot, 3),
      "| p-value:", round(p_value_t2_shoot, 5))
```

```python
print("Control mean:", df["Control_Shoot_Length"].mean())
print("T1 mean:", df["T1_Shoot_Length"].mean())
print("T2 mean:", df["T2_Shoot_Length"].mean())
```

```python

# Control vs T1 (Root Length)
t_stat_t1_root, p_value_t1_root = stats.ttest_ind(
    df["Control_Root_Length"],
    df["T1_Root_Length"],
    equal_var=False,
    nan_policy='omit'
)

# Control vs T2 (Root Length)
t_stat_t2_root, p_value_t2_root = stats.ttest_ind(
    df["Control_Root_Length"],
    df["T2_Root_Length"],
    equal_var=False,
    nan_policy='omit'
)

print("\nRoot Length Comparison")
print("Control vs T1 → t-stat:", round(t_stat_t1_root, 3),
      "| p-value:", round(p_value_t1_root, 5))

print("Control vs T2 → t-stat:", round(t_stat_t2_root, 3),
      "| p-value:", round(p_value_t2_root, 5))
```

```python
print("Control mean:", df["Control_Root_Length"].mean())
print("T1 mean:", df["T1_Root_Length"].mean())
print("T2 mean:", df["T2_Root_Length"].mean())
```
